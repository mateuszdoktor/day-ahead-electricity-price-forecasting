```python
import pandas as pd
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent.parent))
from src.config import DATA_DIR
```

```python
investing_data_path = DATA_DIR / "raw" / "investing_com"

data_files_paths = {
    path.stem: path
    for path in investing_data_path.iterdir()
    if path.is_file() and path.suffix == ".csv"
}
print(data_files_paths.keys())
```

# Investing.com

**Investing.com** is a global financial data platform that offers access to real-time and historical data across international exchanges and financial instruments.

## Description of data from the Investing.com platform

| Category | Data | File Name | Unit | Market Time Unit (MTU) |
|---|---|---|---|---|
| **Energy Commodities & Emissions** | Brent Oil (LCOc1) | `brent_oil_futures.csv` | USD | 1 day |
|| Carbon Emissions (CFI2Zc1) | `carbon_emissions_futures.csv` | EUR | 1 day |
|| Rotterdam Coal (ATWMc1) | `rotterdam_coal_futures.csv` | USD | 1 day |
|| ICE Dutch TTF Natural Gas (TFMBMc1) | `dutch_ttf_natural_gas_futures` | EUR | 1 day |
|| German Power Baseload Electricity Month (DBEc1) | `german_power_baseload_futures.csv` | EUR | 1 day |
| **Foreign Exchange (FX) Rates** | EUR/PLN - Euro Polish Zloty | `eur_pln.csv` | PLN | 1 day |
|| EUR/USD - Euro US Dollar | `eur_usd.csv` | USD | 1 day |

## 1. Energy Commodities & Emissions

### 1.1 Brent Oil (LCOc1)

International benchmark price for crude oil. Data represents the market price of Brent crude oil.

```python
brent_oil = pd.read_csv(data_files_paths.get("brent_oil_futures"))
brent_oil.head(10)
```

### 1.2 Carbon Emissions (CFI2Zc1)

Market price of EU carbon emission allowances. The data reflects the cost of emitting one tonne of CO2 under the EU ETS.

```python
carbon_emissions = pd.read_csv(data_files_paths.get("eu_carbon_emissions_futures"))
carbon_emissions.head(10)
```

### 1.3 Rotterdam Coal (ATWMc1)

Benchmark price for thermal coal traded in the Rotterdam market. The data represents the market price of coal used primarily for electricity generation.

```python
rotterdam_coal = pd.read_csv(data_files_paths.get("rotterdam_coal_futures"))
rotterdam_coal.head(10)
```

### 1.4 Dutch TTF Natural Gas (TFMBMc1)

European benchmark price for natural gas traded at the Title Transfer Facility (TTF) in the Netherlands. The data reflects wholesale natural gas prices in Europe.

```python
dutch_ttf_natural_gas = pd.read_csv(
    data_files_paths.get("dutch_ttf_natural_gas_futures")
)
dutch_ttf_natural_gas.head(10)
```

### 1.5 German Power Baseload Electricity Month (DBEc1)

European benchmark price for German baseload electricity traded in the wholesale power market. The data reflects monthly electricity prices for delivery in Germany.

```python
german_power_baseload = pd.read_csv(
    data_files_paths.get("german_power_baseload_futures")
)
german_power_baseload.head(10)
```

## 2. Foreign Exchange (FX) Rates

### 2.1 EUR/PLN - Euro Polish Zloty

Foreign exchange rate representing the value of the euro in Polish zloty. The data shows how many PLN are required to purchase one EUR.

```python
eur_pln_fx = pd.read_csv(data_files_paths.get("eur_pln"))
eur_pln_fx.head(10)
```

### 2.2 EUR/USD - Euro US Dollar

Foreign exchange rate representing the value of the euro in US dollars. The data shows how many USD are required to purchase one EUR.

```python
eur_usd_fx = pd.read_csv(data_files_paths.get("eur_usd"))
eur_usd_fx.head(10)
```

# Data to be fetched from the ENTSO-E Transparency Platform

**All data and variables fetched from ENTSO-E TP are explained in the [ENTSO-E data overview](./01_ENTSOE_data_overview.ipynb).**

Poland is strongly integrated with the European electricity market, so its market conditions are influenced not only by domestic factors but also by developments in neighbouring and interconnected markets. Selected external markets are therefore included as explanatory variables based on their geographical proximity and relevance to the Polish market.

Poland (PL) is the primary market of the dataset, and all data types listed below will be fetched for Poland, together with its actual load and actual generation per production type. For the other markets, only the data types indicated below will be collected.

The selected data include:

- **Day-ahead prices (P)** 
- **Load forecasts (LF)** 
- **Wind and solar forecasts (WS)**
- **Generation forecasts (GF)**

The data for the whole period (2022-06-09 to 2026-10-21) was downloaded in three ranges (2022-06-09 to 2023-01-01, 2022-12-01 to 2026-09-16 and 2026-09-01 to 2026-10-21), which are merged into one continuous range in the next notebook (`02b_merging_ranges`).

Outages, unit unavailability and hydro storage are not fetched, and the zones AT, FR, HU, NO and SE are not part of the dataset.

```python
import pandas as pd
import sys
import logging

from pathlib import Path

sys.path.append(str(Path.cwd().parent.parent.parent))

from src.api.entsoe_api import EntsoeClient
from src.config import DATA_DIR, DOWNLOAD_START_DATE, DOWNLOAD_END_DATE
from src.utils.file_io import inspect_entsoe_files

logging.basicConfig(level=logging.INFO)
```

```python
client = EntsoeClient()

start_date = pd.Timestamp(DOWNLOAD_START_DATE, tz="Europe/Warsaw")
end_date = pd.Timestamp("2023-01-01", tz="Europe/Warsaw")
date_range = f"{DOWNLOAD_START_DATE}_{DOWNLOAD_END_DATE}"

params = {"start_date": start_date, "end_date": end_date, "save_to_parquet": True}

bidding_zones_data = {
    "DE_LU": [
        "day_ahead_prices",
        "load_forecast",
        "generation_wind_solar_forecast",
        "generation_forecast",
    ],
    "CZ": [
        "day_ahead_prices",
        "load_forecast",
        "generation_wind_solar_forecast",
        "generation_forecast",
    ],
    "SK": ["day_ahead_prices", "load_forecast", "generation_forecast"],
    "LT": [
        "day_ahead_prices",
        "load_forecast",
        "generation_wind_solar_forecast",
        "generation_forecast",
    ],
    "SE_4": [
        "day_ahead_prices",
        "load_forecast",
        "generation_wind_solar_forecast",
        "generation_forecast",
    ],
    # "AT": [
    #     "day_ahead_prices",
    #     "load_forecast",
    #     "generation_wind_solar_forecast",
    #     "generation_forecast",
    # ],
    # "HU": [
    #     "day_ahead_prices",
    #     "load_forecast",
    #     "generation_wind_solar_forecast",
    #     "generation_forecast",
    # ],
    # "FR": [
    #     "day_ahead_prices",
    #     "load_forecast",
    #     "generation_wind_solar_forecast",
    #     "generation_forecast",
    # ],
}
```

```python
for bidding_zone in bidding_zones_data.keys():
    client.fetch(bidding_zone, datasets=bidding_zones_data[bidding_zone], **params)
```

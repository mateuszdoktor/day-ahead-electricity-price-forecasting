# Investing.com Data Cleaning

### All datasets

* Keep only `Price`; drop `Vol.` (empty or partly empty), `Open`, `High`, `Low` (redundant with `Price`; in `german_power_baseload_futures` practically identical to it), `Change %` (derived from `Price`) and `Date` (duplicates the index).
* No deduplication or timezone conversion is required.

```python
import sys
from pathlib import Path

import pandas as pd
import logging

sys.path.append(str(Path.cwd().parent.parent.parent))
from src.config import DATA_DIR, DOWNLOAD_START_DATE, DOWNLOAD_END_DATE

from src.utils.data_quality import (
    print_missing_values,
)

from src.utils.cleaning import clean_investing_com_datasets
from src.utils.file_io import inspect_investing_files, load_investing_data, save_data
```

```python
date_range = f"{DOWNLOAD_START_DATE}_{DOWNLOAD_END_DATE}"
logging.basicConfig(level=logging.INFO)
```

```python
data_inv = load_investing_data(
    DATA_DIR / "raw" / "investing_com",
)
```

```python
data_inv = clean_investing_com_datasets(data_inv)
```

## 1. Missing Values – Confirmation

`Vol.` was the only column with missing values, so after the cleaning none of the five datasets contains NaN values. Only the `Price` column is left in each dataset; weekend and holiday gaps and the descending index order are handled later, during preprocessing.

```python
for name, data in data_inv.items():
    print_missing_values(data=data, name=name)
```

## 2. Conclusions

The cleaning decisions from the data quality audit have been applied to all five datasets, and the check above confirms the result. Each dataset now consists of the `Price` column only, indexed by a UTC date.

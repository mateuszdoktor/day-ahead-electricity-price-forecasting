```python
from jao import JaoPublicationToolPandasClient
from dotenv import load_dotenv
import os
import pandas as pd
```

```python
load_dotenv()
jao_api_key = os.getenv("JAO_API_KEY")

client = JaoPublicationToolPandasClient(api_key=jao_api_key)

day = pd.Timestamp("20260801", tz="Europe/Warsaw")
start = pd.Timestamp("20260701", tz="Europe/Warsaw")
end = pd.Timestamp("20260802", tz="Europe/Warsaw")
```

# JAO - Joint Allocation Office

The Joint Allocation Office (JAO) is a centralized service company and Europe's Single Allocation Platform that auctions cross-border electricity transmission capacity rights on behalf of transmission system operators (TSOs)

## Description of Data from the JAO API
The following table provides an overview of the main variables obtained from the JAO API. The selection and terminology are based primarily on the official [CORE DA Publication Handbook](https://publicationtool.jao.eu/PublicationHandbook/Core_PublicationTool_Handbook_v2.2.pdf)

| Data | API Method Name | Unit | Market Time Unit (MTU) |
|---|---|---|---|
| Max Net Positions | `query_minmax_np` | MW | 1 hour |
| Max Exchanges (Maxbex) | `query_maxbex` | MW | 1 hour |
| Allocation Constraint | `query_allocationconstraint` | MW | 15 min |
| DC2F | `query_dc2f`| MW | 1 hour |


### 1. Max Net Positions

**Description:**

Minimum and maximum Core net positions in MW of each hub for each MTU of the day. These indicators are extracted from the final flow-based domain and final bilateral exchange restriction.

**Publication time:** 

10:30 am (D-1)

```python
minmax_np = client.query_minmax_np(day=day)
minmax_np[[x for x in minmax_np.columns if "PL" in x]]
```

### 2. Max Exchanges (Maxbex)

**Description:**

Maximum bilateral exchanges between two Core hubs with assumption that the other net positions are zero. These indicators are extracted from the final flow-based domain and final bilateral exchange restriction.

**Publication time:**

10:30 am (D-1)


```python
maxbex = client.query_maxbex(day=day, from_zone="PL")
maxbex
```

### 3. Allocation Constraints 

**Description:**

As per the Core CCM, Belgium, Poland and the Netherlands are allowed to use external constraints.

**Publication time:**

10:30 am (D-1)

```python
allocationconstraints = client.query_allocationconstraint(d_from=start, d_to=end)
allocationconstraints_pl = allocationconstraints[
    [x for x in allocationconstraints.columns if "PL" in x]
]
allocationconstraints_pl
```

### 4. DC2F

**Description:**

Publishes the aggregated assumptions from the grid models for each MTU on TSU and Hub level: Vertical load, generation(production) and net position for each Core hub and TSO if it differs from the hub level
- “Vertical load” is the load as seen from the transmission grid in MW in the Common Grid Model
- “Generation” is the generation in MW in the Common Grid Model
- “Core net position” is the forecast of the overall balance of the countries in MW in the Common Grid Model

**Publication time:**

10:30 am (D-1)

```python
dc2f = client.query_d2cf(d_from=start, d_to=end)
countries = ["PL", "DE", "CZ", "SK", "AT", "HU"]
countries_cols = [x for x in dc2f.columns if any(c for c in countries if c in x)]
dc2f[countries_cols]
```
